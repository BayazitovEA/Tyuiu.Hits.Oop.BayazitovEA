<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создан базовый класс Employee на C#, представляющий информацию о сотрудниках компании. На основе этого класса разработаны 3 производных класса, демонстрирующих принципы наследования и полиморфизма: Manager, Developer и Designer. В каждом из классов реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;


public class Employee
{

    public string Name { get; set; }
    public int Age { get; set; }
    public double Salary { get; set; }


    public Employee(string name, int age, double salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }


    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Зарплата: {Salary}";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет общие рабочие обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }
}




public class Manager : Employee
{
    public int SubordinatesCount { get; set; }

    // Конструктор с использованием base и сеттера
    public Manager(string name, int age, double salary, int subordinatesCount) 
        : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} управляет командой из {SubordinatesCount} человек.");
    }

    // Новый метод — взаимодействие с другими объектами
    public void AssignTask(Employee employee, string task)
    {
        Console.WriteLine($"{Name} (менеджер) поручает задачу '{task}' сотруднику {employee.Name}.");
        employee.Work(); // Вызов метода другого объекта
    }
}


public class Developer : Employee
{
    public string Specialization { get; set; }

    public Developer(string name, int age, double salary, string specialization) 
        : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} (Разработчик) уходит в отпуск. Необходимо согласовать сроки сдачи проектов.");
    }

    // Новый метод — взаимодействие
    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код на {Specialization}.");
    }
}


public class Designer : Employee
{
    public string DesignStyle { get; set; }

    public Designer(string name, int age, double salary, string designStyle) 
        : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} работает над дизайн-проектом в стиле {DesignStyle}.");
    }

    // Новый метод — взаимодействие
    public void CreateMockup()
    {
        Console.WriteLine($"{Name} создает макет в стиле {DesignStyle}.");
    }
}



Console.WriteLine("--- Информация о сотрудниках ---");


Manager manager = new Manager("Игорь", 45, 150000, 10);
Developer developer = new Developer("Анна", 28, 120000, "Backend");
Designer designer = new Designer("Петр", 32, 90000, "Минимализм");


Console.WriteLine(manager.GetInfo());
manager.Work();
manager.TakeVacation();

Console.WriteLine();

Console.WriteLine(developer.GetInfo());
developer.Work();
developer.TakeVacation();

Console.WriteLine();

Console.WriteLine(designer.GetInfo());
designer.Work();
designer.TakeVacation();


Console.WriteLine("\n--- Взаимодействие объектов ---");

// Менеджер поручает задачи другим сотрудникам
manager.AssignTask(developer, "Разработать API");
manager.AssignTask(designer, "Нарисовать логотип");

// Прямое взаимодействие через методы
developer.WriteCode();
designer.CreateMockup();
```
